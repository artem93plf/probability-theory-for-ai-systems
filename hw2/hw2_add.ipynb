### Тема 1. Моделирование геометрической вероятности
> Смоделируйте вычисление геометрической вероятности (на примере из лекции про метание дротиков в игровой круг, площадь которого в два раза меньше площади стены). Показать теоретический расчет и графическую иллюстрацию для рассматриваемого примера.


```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, Polygon
import numpy as np
```

```python
def points(n):
    a = 12
    fig, ax = plt.subplots(figsize=(5, 5))
    ax = plt.gca()

    triangle = Polygon([[0,0], [a, 0], [a / 2, a]])
    ax.add_patch(Rectangle((0, 0), 12, 12, facecolor='g'))
    ax.add_patch(triangle)
    X = np.random.uniform(0, a, n)
    Y = np.random.uniform(0, a, n)
    plt.scatter(X, Y, marker='.', c="black")
    inside = []
    for x, y in zip(X, Y):
        if y <= 2*x and y <= -2*x + 2 * a:
            inside.append([x, y])
    prob = len(inside) / n
    

    plt.show()
    print(f"в синий треугольник из {n} точек попало {len(inside)}. Доля попавших = {prob}")
```

```python
points(100)
```
